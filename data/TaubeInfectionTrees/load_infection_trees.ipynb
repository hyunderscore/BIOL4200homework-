```python
import networkx as nx
import numpy as np
import polars as pl
import xgi
```

```python
df_edges = pl.read_csv("edgelists.csv")
df_nodes = pl.read_csv("node_info.csv")
tree_names = df_nodes["id"].unique().to_list()
tree_names
```

```python
trees = {}
for t in tree_names:
    G = nx.DiGraph()
    G.add_nodes_from(df_nodes.filter(df_nodes["id"] == t)["name"].to_list())
    edges = df_edges.filter(df_edges["id"] == t)
    G.add_edges_from(zip(edges["from"].to_list(), edges["to"].to_list()))
    G.name = t
    trees[t] = G
```

```python
trees = {}
for t in tree_names:
    G = xgi.Hypergraph()
    G.add_nodes_from(df_nodes.filter(df_nodes["id"] == t)["name"].to_list())
    edges = df_edges.filter(df_edges["id"] == t)
    G.add_edges_from(zip(edges["from"].to_list(), edges["to"].to_list()))
    G.name = t
    trees[t] = G
```

```python
np.argmax([trees[t].number_of_edges() for t in tree_names])
```
